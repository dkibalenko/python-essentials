Reference Counting

```python
class Animal:
    def __init__(self, name: str):
        self.name = name
        

cat = Animal("Cat1")
```

```python
import sys

sys.getrefcount(cat)
```

```python
pet = cat
sys.getrefcount(cat)
```

```python
the_same_cat = cat
sys.getrefcount(pet)
```

```python
another_cat = Animal("Cat2")
sys.getrefcount(another_cat)
```

if we instantiate an object with no reference and pass it directly into the `sys.getrefcount()` we will get 1:

```python
sys.getrefcount(Animal("Cat3"))
```

### Garbage Collector

```python
import sys


class Animal:
    def __init__(self, name: str):
        self.name = name
```

```python
cat = Animal("Cat1")
pet = cat
the_same_cat = cat

sys.getrefcount(cat)
```

```python
del the_same_cat
sys.getrefcount(cat)
```

```python
del pet
sys.getrefcount(cat)
```

```python
del cat
sys.getrefcount(cat)
```

if we delete all the references to the object, it gets deleted by ***garbage collector***. But if we want be informed about it, we can implement `__del__` method in out custom class with some information about it:

```python
import sys


class Animal:
    def __init__(self, name: str):
        self.name = name
        
    def __del__(self):
        print(self.name, "deleted")
```

```python
cat = Animal("Cat1")
pet = cat
the_same_cat = cat
```

```python
sys.getrefcount(cat)
```

then we delete references to the object one by one:

```python
del the_same_cat
```

```python
del pet
```

```python
del cat
```

### Shallow vs Deep Copy

no copy, only shared reference:

```python
class Candy:
    def __init__(self, taste: str):
        self.taste = taste
```

```python
chocolate = Candy("Chocolate")
lemon = Candy("Lemon")

box1 = [chocolate, lemon]
```

```python
box2 = box1  # shared reference

box2[0] = Candy("Caramel")

box1[0].taste
```

### Shallow copy

```python
chocolate = Candy("Chocolate")
lemon = Candy("Lemon")

box1 = [chocolate, lemon]
```

```python
box2 = box1.copy()  # shallow copy

box2[0] = Candy("Caramel")

box1[0].taste
```

```python
box2[1].taste = "Orange"  # since 'box2' is made of shallow copy of 'box1' it has shared reference to the objects inside 'box1'

box1[1].taste
```

deepcopy with `deepcopy()`

```python
chocolate = Candy("Chocolate")
lemon = Candy("Lemon")

box1 = [chocolate, lemon]
```

```python
from copy import deepcopy
```

```python
box2 = deepcopy(box1)  # deepcopy

box2[1].taste = "Orange"

box1[1].taste
```
