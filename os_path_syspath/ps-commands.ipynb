### Env Vars
1. View all environment variables
```bash
Get-ChildItem Env:
   
Name                           Value                                                                                                                                                                      
----                           -----                                                                                                                                                                      
_OLD_VIRTUAL_PATH              c:\Users\Dennie\AppData\Roaming\Code\User\globalStorage\github.copilot-chat\debugCommand;c:\Users\Dennie\AppData\Roaming\Code\User\globalStorage\github.copilot-chat\cop...
ALLUSERSPROFILE                C:\ProgramData                                                                                                                                                             
APPDATA                        C:\Users\Dennie\AppData\Roaming        
...
```
2. View a specific variable
```bash
$env:PATH
C:\Users\Dennie\Mate Academy\Python Core\python_essentials\.venv\Scripts;
...
```
3. View machine vs user PATH separately
```bash
 [Environment]::GetEnvironmentVariable("Path","Machine")
C:\Python313\Scripts\;C:\Python313\;C:\Program Files (x86)\Common Files\Oracle\Java\javapath;C:\WINDOWS\system32;C:\WINDOWS;C:\WINDOWS\System32\Wbem;
...


[Environment]::GetEnvironmentVariable("Path","User")
c:\python313\Scripts;C:\Python313\Scripts\;C:\Python313\;C:\Program Files (x86)\Common Files\Oracle\Java\javapath;C:\WINDOWS\system32;C:\WINDOWS;C:\WINDOWS\System32\Wbem;
```
> PowerShell supports process, user, and machine scopes.
4. Set variable for current session (process scope)
```bash
$env:MYVAR = "value"
```
This persists only until the session closes.

5. Persist variable for the current user
```bash
[Environment]::SetEnvironmentVariable("MYVAR", "value", "User")
```
6. Persist variable for the entire machine (requires admin)
```bash
[Environment]::SetEnvironmentVariable("MYVAR", "value", "Machine")
```
7. Delete environment variables
```bash
Remove-Item Env:MYVAR
```

#### 📁 Filesystem ([link](https://learn.microsoft.com/en-us/powershell/module/microsoft.powershell.management/?view=powershell-7.6))
- `Get-ChildItem` — list files/directories\

- `Set-Location` — change directory

- `Copy-Item` — copy files

- `Move-Item` — move files

- `Remove-Item` — delete files

- `New-Item` — create files/folders

#### 🔐 File permissions ([link](https://learn.microsoft.com/en-us/powershell/module/microsoft.powershell.security/?view=powershell-7.6))

- `Get-Acl` — view permissions
    
- `Set-Acl` — modify permissions (Part of core PowerShell filesystem provider)
    

#### 🖥️ OS & system commands info ([link](https://learn.microsoft.com/en-us/powershell/module/microsoft.powershell.management/))

- `Get-Process`
    
- `Get-Service`
    
- `Start-Service` / `Stop-Service`
    
- `Get-ComputerInfo`
    
- `Get-WmiObject` / `Get-CimInstance`
    

#### 🌐 Networking & HTTP (curl equivalent [link](https://learn.microsoft.com/en-us/powershell/module/microsoft.powershell.utility/?view=powershell-7.6))

PowerShell uses `Invoke-WebRequest` and `Invoke-RestMethod` instead of curl:

- `Invoke-WebRequest` — raw HTTP requests
    
- `Invoke-RestMethod` — JSON API requests
    

Example:
```bash
Invoke-WebRequest -Uri "https://example.com"
Invoke-RestMethod -Method POST -Uri "https://api.example.com" -Body @{a=1}
```
