```python
# the import cache
import sys

"math" in sys.modules  # but, 'math' is still fully loaded and stored in sys.modules['math']
```

`sys.modules` maps: `module_name → module_object`

Why it matters
- Prevents re‑executing module code
- Makes imports fast
- Allows circular imports to partially work
- You can inspect what is already loaded
