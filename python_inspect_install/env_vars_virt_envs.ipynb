### 🌐 Sending HTTP Requests (PowerShell’s “curl”)
PowerShell aliases `curl` to `Invoke-WebRequest`, but the real command is:

**GET request**

```python
Invoke-WebRequest https://example.com
```

**POST request**

```python
Invoke-WebRequest -Uri https://example.com/api -Method POST -Body '{"a":1}' -ContentType "application/json"
```

**Download a file**

```python
Invoke-WebRequest -Uri https://example.com/file.zip -OutFile file.zip
```
