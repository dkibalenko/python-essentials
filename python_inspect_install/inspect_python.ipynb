### 🪟 Windows (PowerShell or CMD)
#### 1. Using the Python Launcher (BEST method)
The Python Launcher (`py.exe`) can list every Python installation it knows about.

**List all installed Python versions**:

```python
py -0
 -V:3.14 *        Python 3.14 (64-bit)
 -V:3.13          Python 3.13 (64-bit)
 -V:3.12          Python 3.12 (64-bit)
 -V:3.11          Python 3.11 (64-bit)
 -V:3.10          Python 3.10 (64-bit)
 -V:ContinuumAnalytics/Anaconda39-64 Anaconda 2022.10
```

**List versions with full paths**

```python
py -0p
 -V:3.14 *        C:\Program Files\Python314\python.exe
 -V:3.13          C:\Python313\python.exe
 -V:3.12          C:\Users\Dennie\AppData\Local\Programs\Python\Python312\python.exe
 -V:3.11          C:\Users\Dennie\AppData\Local\Programs\Python\Python311\python.exe
 -V:3.10          C:\Python\python.exe
 -V:ContinuumAnalytics/Anaconda39-64 C:\Users\Dennie\anaconda3\python.exe
```

This is the most accurate method on Windows.

---

#### 2. Using PowerShell to find all `python.exe` files

```python
Get-Command python
   
CommandType     Name                                               Version    Source                                                                                 
-----------     ----                                               -------    ------                                                                                 
Application     python.exe                                         3.14.15... C:\Program Files\Python314\python.exe 
```

This lists all python executables found in `PATH`.

You can also search for all python executables:

```python
Get-Command python.*

CommandType     Name                                               Version    Source                                                                                 
-----------     ----                                               -------    ------                                                                                 
Application     python.exe                                         3.14.15... C:\Program Files\Python314\python.exe                                                  
Application     python.exe                                         3.13.11... C:\Python313\python.exe  
```
