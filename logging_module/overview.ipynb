### What each notebook covers

**`01_basics.ipynb` — the mental model**

|Cell|Topic|
|---|---|
|b001|Why not `print()`? — the log record flow diagram|
|b002|Five levels, numeric values, decision rule for each|
|b003|`logging.basicConfig()` — simplest possible setup, `force=True` for Jupyter|
|b004|The singleton — `getLogger(__name__)` returns the same object; the name hierarchy|
|b005|Live singleton demo — `is`, `.parent`, root logger|
|b006|Handlers, Formatters, full `LogRecord` attribute table|
|b007|Manual setup — Logger → Handler → Formatter, the `if not logger.handlers:` guard|
|b008|Propagation — how records bubble up, the double-logging trap explained|
|b009|Live double-logging demo + both fixes (`propagate=False` and removing the handler)|

**`02_advanced.ipynb` — production config**

|Cell|Topic|
|---|---|
|a001|Why `dictConfig()` beats manual setup — atomic, idempotent, separable|
|a002|`dictConfig()` inline dict — console + RotatingFileHandler, `ext://`, writes to `logs/app.log`|
|a003|JSON config file pattern — directory structure, field-by-field reference, `ext://` explained|
|a004|Full demo: write `logging_config.json` → load it → apply → verify file output|
|a005|`__name__` pattern for every module + `NullHandler` for library code and why|
|a006|Module hierarchy demo — effective levels, library logger filtered by root|
|a007|`LoggerAdapter` for per-request context (`request_id`) vs `Filter` for global context|
|a008|Live request adapter demo — two simulated HTTP requests with distinct `request_id`s|
|a009|Four gotchas + decision tree: basicConfig → script, dictConfig dict → medium project, JSON config → production service, NullHandler → library|
