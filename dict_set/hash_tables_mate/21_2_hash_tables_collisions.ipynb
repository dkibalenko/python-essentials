but, when we **overload** `__eq__` for our custom class:

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width
        
    def __repr__(self):
        return f"Rectangle(length={self.length}, width={self.width})"
    
    def __eq__(self, other):
        return self.length == other.length and self.width == other.width
    
    def __hash__(self):
        return self.length + self.width
```

```python
r1 = Rectangle(10, 20)
r2 = Rectangle(10, 20)
```

```python
r1 == r2
```

```python
hash(r1)
```

```python
hash(r2)
```

there is no collision still, it would be one `if r1 != r2 == True` and the **hash was the same**

### Collisions occur when two objects are not equal, but their hashes are the same

### The Golden Rule of Hashing
If `a == b` is `True`, then `hash(a)` must equal `hash(b)`

Your `__hash__` function is "bad" because it can produce the same hash for unequal objects (`10 + 20` and `15 + 15` both equal `30`). This isn't technically an error, but it leads to a hash collision and makes the dictionary less efficient.

```python
rectangles = {}
```

```python
rectangles[r1] = 100
rectangles[r2]
```

since **hashes are the same**, key `r2` would return the **same value**

### Overwriting one dictionary key by another one

Next, when using objects that **equal to** each other and have the **same hash**, they will be considered the **same key in a dictionary**. This means that if you try to use **both instances as keys** in the same dictionary, the second instance will **overwrite** the value associated with the first instance.

```python
rectangles[r2] = 200
rectangles
```

if keys are the same, the value will be overwritten

### Collision

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width
        
    def __repr__(self):
        return f"Rectangle(length={self.length}, width={self.width})"
    
    def __eq__(self, other):
        return self.length == other.length and self.width == other.width
    
    def __hash__(self):
        return self.length + self.width


rectangles = {}
r1 = Rectangle(10, 20)
rectangles[r1] = 100
rectangles
```

```python
r2 = Rectangle(15, 15)
```

```python
r1 == r2
```

```python
hash(r1)
```

```python
hash(r2)
```

```python
# collision (since object is not equal to object but they have the same hash value)
hash(r1) == hash(r2)
```

```python
rectangles[r2] = 200
```

```python
rectangles
```

objects are unequal and they have the same hash value, but they all get added to the dict

```python
r3 = Rectangle(10, 4)
```

```python
r1 == r3
```

```python
hash(r1) == hash(r3)
```

```python
rectangles[r3] = 250
```

```python
rectangles
```

```python
hash(r3)
```
