### `hash`

`O(1)` time complexity in all cases

```python
hash(1)
```

```python
hash(10)
```

```python
hash(10.1)
```

for the ***same values*** it returns the same ***hash***

```python
hash("hello world")
```

```python
hash("hello world")
```

we can always get ***hash*** with some value, but ***NOT*** always the other way around

```python
hash([1, 2, 3])
```

by default list and almost all ***mutable*** types are ***UNHASHABLE***, they cannot be hashed (to be added hash value)

```python
class User:
    def __init__(self):
        self.name = "John"
        self.surname = "Smith"
```

```python
hash(User())
```

possible to ***re-implement*** hash func:

```python
class User:
    def __init__(self):
        self.name = "John"
        self.surname = "Smith"
        
    def __hash__(self):
        return 10
```

```python
hash(User())
```

it's not good way to reimplement hash function

```python
t = (2,)
```

```python
hash(t)
```

```python
t = ([1],)
```

```python
hash(t)
```

Tuples and frozensets are immutable containers and can only be hashable if all their ***members are hashable***
