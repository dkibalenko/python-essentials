```python
import sys


my_dict = {}

for i in range(20):
    my_dict[i] = 100
    print(f"elements = {i+1} size {sys.getsizeof(my_dict)}")
```

```python
# by default, dict is created with size of 8 slots

CAPACITY = 8
```

```python
int(8 * (2 / 3))  # threshold = 5
```

when a dict is filled out up to 2/3 it gets resized twice

the size of dict is 5 elements, when its CAPACITY = 8 (2/3 of it), after passing its threshold, it gets resized

```python
int(16 * (2 / 3))  # threshold = 10
```

after passing next threshold of 10 elements, it gets resized again
