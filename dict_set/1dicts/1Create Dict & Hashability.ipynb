### Creating Python Dictionaries

[Docs - Dictionaries](https://docs.python.org/3/tutorial/datastructures.html#dictionaries)

[Docs - Mapping Types](https://docs.python.org/3/library/stdtypes.html#typesmapping)

[Docs - Hashable & Immutable](https://docs.python.org/3/glossary.html#term-hashable)

Unlike sequences, which are indexed by a range of numbers, dictionaries are indexed by keys, which can be **any immutable type**; strings and numbers can always be keys.

Tuples can be used as keys if they contain **only** strings, numbers, or tuples; if a tuple contains any mutable object either directly or indirectly, it cannot be used as a key. You can’t use lists as keys, since lists can be modified in place using index assignments, slice assignments, or methods like `append()` and `extend()`.

There are different mechanisms available to create dictionaries in Python.

#### Literals

We can use a literal to create a dictionary:

````python
class A:
    pass

a = A()
aa = A()
````

````python
a == aa
````

````python
a == a
````

````python
id(a)
````

````python
id(aa)
````

````python
d = {a: 1, aa: 2}
````

````python
d
````

````python
a = {'k1': 100, 'k2': 200}
````

````python
a
````

Note that the order in which the items are listed in the literal is maintained when listing out the elements of the dictionary. This does not hold for Python version earlier than 3.6 (practically, version 3.5).

Another thing to note is that dictionary **keys** must be hashable objects. Associated values on the other hand can be any object.

So tuples of hashable objects are themselves hashable, but lists are not, even if they only contain hashable elements.

> Tuples of non-hashable elements are also **not hashable**.

````python
hash((1, 2, 3))
````

````python
hash([1, 2, 3])
````

````python
hash(([1, 2], [3, 4]))
````

So we can create dictionaries that look like this:

````python
a = {('a', 100): ['a', 'b', 'c'], 'key2': {'a': 100, 'b': 200}}
````

````python
a
````

Interestingly, functions are **hashable**:

````python
def my_func(a, b, c):
    print(a, b, c)
````

````python
hash(my_func)
````

Which means we can use functions as keys in dictionaries:

````python
d = {my_func: [10, 20, 30]}
````

A simple application of this might be to store the argument values we want to use to call the function at a later time:

````python
def fn_add(a, b):
    return a + b

def fn_inv(a):
    return 1/a

def fn_mult(a, b):
    return a * b
````

````python
funcs = {fn_add: (10, 20), fn_inv: (2,), fn_mult: (2, 8)}
````

Remember that when we iterate through a dictionary we are actually iterating through the **keys**:

````python
for f in funcs:
    print(f)
````

We can then call the functions this way:

````python
for f in funcs:
    result = f(*funcs[f])
    print(result)
````

We can also iterate through the items (as tuples) in a dictionary as follows:

````python
for f, args in funcs.items():
    print(f, args)
````

So we could now call each function this way:

````python
for f, args in funcs.items():
    result = f(*args)
    print(result)
````

#### Using Comprehensions

We can also create dictionaries using a dictionary comprehension.
This is very similar to list comprehensions or generator expressions.

Suppose we have two iterables, one containing some keys, and one containing some values we want to associate with each key:

````python
keys = ['a', 'b', 'c']
values = (1, 2, 3)
````

We can then easily create a dictionary this way - the non-Pythonic way!

````python
d = {}  # creates an empty dictionary
for k, v in zip(keys, values):
    d[k] = v
````

````python
d
````

But it is much simpler to use a dictionary comprehension:

````python
d = {k: v for k, v in zip(keys, values)}
````

````python
d
````

Dictionary comprehensions support the same syntax as list comprehensions - you can have nested loops, `if` statements, etc.

````python
keys = ['a', 'b', 'c', 'd']
values = (1, 2, 3, 4)

d = {k: v for k, v in zip(keys, values) if v % 2 == 0}
````

````python
d
````

In the following example we are going to create a grid of 2D coordinate pairs, and calculate their distance from the origin:

````python
x_coords = (-2, -1, 0, 1, 2)
y_coords = (-2, -1, 0, 1, 2)
````

If you remember list comprehensions, we would create all possible `(x,y)` pairs using nested loops (a Cartesian product):

````python
grid = [
    (x, y) 
    for x in x_coords 
    for y in y_coords
]
grid
````

````python
import math
````

We can use the `math` module's `hypot` function to do calculate these distances

````python
math.hypot(1, 1)
````

So to calculate these distances for all our points we would do this:

````python
grid_extended = [(x, y, math.hypot(x, y)) for x, y in grid]
grid_extended
````

We can now easily tweak this to make a dictionary, where the coordinate pairs are the key, and the distance the value:

````python
grid_extended = {(x, y): math.hypot(x, y) for x, y in grid}
````

````python
grid_extended
````

**Q2 — ❌ Wrong, and this is the interesting one.** Dict literals don't keep duplicate keys — they collapse, last write wins:

```python
d = {greet: 1, greet: 2}

Same mechanism as `{1: 'a', 1: 'b'}` → `{1: 'b'}`. **Q7 — Order-of-operations**

```shell
d = {"a": 1, "b": 2}
d["a"] = 99
d["c"] = 3
print(list(d.keys()))
```
Does reassigning `"a"` change its position in iteration order? Why or why not — tie this to what you know about dict internals.

**Answer**:

Q7:

Reassigning will not change postion in iteration order. Keys are used by hash function to compute a hash value then mod by dict's length to get final array's slot index to store the value. Since Python 3.5 dicts are ordered, so I think `'a'`, `'b'`, `'c'` all maintain the same relative order even though reassigning their values multiple times.

**Q7 — ✅ Right conclusion, refine the internals.** Reassigning doesn't move position — correct. But the mechanism isn't quite "hash mod length decides iteration order."Since 3.6, CPython's dict is actually two structures:
- a sparse hash table holding only _indices_
- plus a dense array holding the actual key/value/hash entries in **insertion order**.
Iteration walks the dense array directly — that's _why_ order is preserved regardless of hash values. The sparse table (where hash-mod-length probing happens) is only used to _find_ an entry, never to decide iteration order. Reassigning `d["a"] = 99` finds the existing dense-array slot and updates the value in place — no new entry, no reorder. Only a genuinely new key appends to the dense array.

**Q8 — Interview-style conceptual**

Your notebook says "dictionary keys must be hashable." State the actual rule more precisely — hashable isn't quite sufficient on its own. What's the real requirement, and what's a hashable object that would still make a bad dict key even though Python allows it?

**Answer**:

The requrement for Python's dict keys is that they must be not only hashable but any immutable. Immutable objects guarantee thier hash value is constant. Hashability makes objects usable as dict keys or set memebers, since they use hash function internally to produce a hash value, they are hash table. No idea about a hashable object that can still be a bad dict key. But talking about the hashable object itself that is accepted but is bad id difficult.

**Q8 — Partial credit, and this is the one to sit with.** Immutability is the _convention_, not the _rule_.
> The actual formal requirement Python enforces is: **objects that compare equal must have equal hashes, and an object's hash must not change while it's used as a key.**

Immutability is just how built-ins guarantee that — nothing stops you from writing a hashable, mutable, badly-behaved key:

````python
class BadKey:
    def __init__(self, items):
        self.items = items
    def __hash__(self):
        return hash(tuple(self.items))
    def __eq__(self, other):
        return self.items == other.items
````

````python
k = BadKey([1, 2, 3])
d = {k: "value"}
hash(k)
````

````python
k.items.append(4)      # mutate after insertion — Python allows this
hash(k)
````

````python
print(d.get(k))        # None — lookup fails, hash changed after placement
````

`k` is 100% hashable — Python never complains — but it's a terrible key because the hash that decided its bucket is now stale. This is exactly the kind of "hashable but broken" example interviewers use to check you understand the _invariant_, not just the vocabulary word "hashable."
