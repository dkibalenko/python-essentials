```python
def is_prime(number: int) -> bool:
    # write your code here
    if number <= 1:
        return False
    for n in range(2, abs(number)):
        if number % n == 0:
            return False
    return True
```

```python
print(is_prime(1)) # False
print(is_prime(2)) # True
print(is_prime(13)) # True
print(is_prime(-3)) # False
```

```python
def is_prime(number: int) -> bool:
    if number < 2:
        return False

    for i in range(2, int(number**0.5) + 1):  # number ** 0.5 is sqrt of the number(math.sqrt())
        if number % i == 0:
            return False
    return True
```

```python
print(is_prime(1)) # False
print(is_prime(2)) # True
print(is_prime(13)) # True
print(is_prime(-3)) # False
```
