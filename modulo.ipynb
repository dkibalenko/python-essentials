### Python Modulo Operator Basics

Modulo Operator With `int`

The **modulo operator** `%`, when used with two **positive integers**, will return the ***remainder*** of standard Euclidean division:

```python
15 % 4
```

```python
17 % 12
```

```python
240 % 13
```

```python
22 % 0
```

Similar to `int`, the modulo operator used with a `float` will return the remainder of division, but as a `float` value:

```python
12.5 % 5.5
```

```python
17.0 % 12.0
```

An alternative to using a `float` with the modulo operator is to use `math.fmod()` to perform modulo operations on `float` values:

```python
import math
math.fmod(12.5, 5.5)
```

```python
math.fmod(8.5, 2.5)
```

The official Python docs suggest using `math.fmod()` over the Python modulo operator when working with `float` values because of the way `math.fmod()` calculates the result of the modulo operation.

Just like other arithmetic operators, the modulo operator and `math.fmod()` may encounter rounding and precision issues when dealing with floating-point arithmetic:

```python
13.3 % 1.1
```

```python
import math
math.fmod(13.3, 1.1)
```

If maintaining floating-point precision is important to your application, then you can use the modulo operator with `decimal.Decimal`.

* `0 -> 3 -> 2 -> 1` one round, then stop at 1

* `0 -> -3 -> -2 -> -1`

### Modulo Operator and `divmod()`

Takes two parameters and returns a `tuple` containing the results of floor division and modulo using the supplied parameters.

```python
divmod(37, 5)
```

```python
37 // 5
```

```python
37 % 5
```

when the modulo operator is used with an `int`, the remainder will take the sign of the ***divisor***:

```python
divmod(37, -5)
```

```python
37 // -5
```

```python
37 % -5
```

#### Modulo Operator Precedence

The modulo operator `%` shares the same level of precedence as the multiplication `*`, division `/`, and floor division `//` operators.

```python
4 * 10 % 12 - 9
```

### Python Modulo Operator in Practice

How to Check if a Number Is Even or Odd

```python
def is_even(num):
    return num % 2 == 0
```

```python
def is_odd(num):
    return num % 2 != 0
```

```python
is_even(5), is_even(4)
```

```python
is_odd(5), is_odd(4)
```

How to Run Code at Specific Intervals in a Loop

```python
def split_names_into_rows(name_list, modulus=3):
    for index, name in enumerate(name_list, start=1):
        print(f"{name:-^15} ", end="")
        if index % modulus == 0:
            print()
    print()
```

```python
names = ["Picard", "Riker", "Troi", "Crusher", "Worf", "Data", "La Forge"]
split_names_into_rows(names)
```

```python
split_names_into_rows(names, modulus=4)
```

```python
split_names_into_rows(names, modulus=2)
```

```python
split_names_into_rows(names, modulus=1)
```
