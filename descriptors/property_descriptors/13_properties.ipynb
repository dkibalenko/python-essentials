### Getters and Setters

```python
class GlassOfWater:
    def __init__(self, temperature):
        self.temperature = temperature
        
glass = GlassOfWater(temperature=20)
```

```python
glass.temperature = 15
glass.temperature
```

using ***getter*** and ***setter*** of the attribute:

```python
class GlassOfWater:
    def __init__(self, temperature):
        self._temperature = temperature  # protected
        
    # get attribute value
    def get_temperature(self):
        return self._temperature
    
    # set attribute value
    def set_temperature(self, value):
        if not (0 <= value <= 100):
            raise ValueError("Temperature must be within range of 0 - 100.")
        self._temperature = value
        
glass = GlassOfWater(temperature=20)
```

```python
glass.set_temperature(25)
```

```python
glass.get_temperature()
```

```python
try:
    glass.set_temperature(10000)
except ValueError as ex:
    print(ex)
```

### Properties

```python
class GlassOfWater:
    def __init__(self, temperature):
        self._temperature = temperature

    def get_temperature(self):
        print("Get temperature")
        return self._temperature

    def set_temperature(self, value):
        print("Set temperature")
        if not (0 <= value <= 100):
            raise ValueError("Temperature must be within range of 0 - 100.")
        self._temperature = value
        
    temperature = property(fget=get_temperature, fset=set_temperature)
```

```python
glass = GlassOfWater(temperature=20)
```

```python
glass.temperature = 25
```

```python
glass.temperature = 30
```

```python
glass.temperature
```

```python
glass.temperature = 1000
```

### `@property`  Decorator

```python
class GlassOfWater:
    def __init__(self, temperature):
        self._temperature = temperature
        
    @property
    def temperature(self):
        print("Get temperature")
        return self._temperature
    
    @temperature.setter
    def temperature(self, value):
        print("Set temperature")
        if not (0 <= value <= 100):
            raise ValueError("Temperature must be within range of 0 - 100.")
        self._temperature = value
```

```python
glass = GlassOfWater(temperature=20)
```

```python
glass.temperature = 25
```

```python
glass.temperature
```

### use ***property*** in `__init__`

```python
class GlassOfWater:
    def __init__(self, temperature):
        self._temperature = temperature  # there is an assignment bypassing 'setter' validation test
        
    @property
    def temperature(self):
        print("Get temperature")
        return self._temperature
    
    @temperature.setter
    def temperature(self, value):
        print("Set temperature")
        if not (0 <= value <= 100):
            raise ValueError("Temperature must be within range of 0 - 100.")
        self._temperature = value
```

```python
glass = GlassOfWater(temperature=1000)
```

```python
glass.temperature
```

so there was not validation via ***setter*** method

```python
class GlassOfWater:
    def __init__(self, temperature):
        self.temperature = temperature  # instantiating class by passing 'setter' validation
        
    @property
    def temperature(self):
        print("Get temperature")
        return self._temperature
    
    @temperature.setter
    def temperature(self, value):
        print("Set temperature")
        if not (0 <= value <= 100):
            raise ValueError("Temperature must be within range of 0 - 100.")
        self._temperature = value
```

```python
glass = GlassOfWater(temperature=1000)
```

this time we get an exception as expected

### Read-only attributes

```python
class Company:
    def __init__(self, name):
        self.name = name
        
company = Company("Mate Academy")
```

```python
company.name
```

```python
company.name = "mate academy"
```

```python
company.name
```

we can restrict access to the attribute:

```python
class Company:
    def __init__(self, name):
        self._name = name
        
    @property
    def name(self):
        return f"This is read_only: '{self._name}'!"
        
company = Company("Mate Academy")
```

```python
company.name
```

```python
company.name = "mate academy"
```
