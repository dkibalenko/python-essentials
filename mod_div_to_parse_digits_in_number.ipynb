### test digits in a number: if neigbor digit differs more then one it is False, otherwise is True

```python
# get decimal, hundred, thousand ... remainder of a number:

print(78 % 10)
print(789 % 100)
print(4569 % 1000)
```

```python
# with div //, we can get hundred, thousand.. whole part of a number:
print(87 // 10)
print(394 // 10)
print(4885 // 10)
```

```python
def is_jumping(number: int) -> str:
    prev_digit = number % 10  # get number's decimal part (or % 100 if hundred, and so on...)
    number = number // 10  # get number's whole part from decimal to left
    while number > 0:
        if abs(prev_digit - number % 10) != 1:  # if absolute value is greate than 1 it's False
            return "NOT JUMPING"
        prev_digit = number % 10  # if absolute value is 1, again get right most digit...
        number = number // 10  # ... and the rest of left most part of the number
    return "JUMPING"
```

```python
print(is_jumping(0))  # 'JUMPING'
print(is_jumping(77))  # NOT JUMPING
print(is_jumping(9))  # "JUMPING"
print(is_jumping(79))  # 'NOT JUMPING'
print(is_jumping(7889))  # NOT JUMPING
print(is_jumping(23454))  # JUMPING
print(is_jumping(12543))  # NOT JUMPING
print(is_jumping(123453))  # NOT JUMPING
print(is_jumping(79876567))  # NOT JUMPING
```

### check if any digit of the number does fit condition:

```python
def is_special_number(number: int) -> str:
    # write your code here
    while number:
        last_digit = number % 10  # get last digit
        number = number // 10  # get the rest of number to the left from the last one
        if last_digit > 5:
            return 'NOT!!'
    return 'Special!!'
```

```python
print(is_special_number(2))  # "Special!!"
print(is_special_number(9))  # "NOT!!"
print(is_special_number(23))  # "Special!!"
print(is_special_number(38))  # "NOT!!"
```

```python
def is_special_number(number: int) -> str:
    while number > 0:
        if number % 10 > 5:
            return "NOT!!"
        number //= 10
    return "Special!!"
```

### test if digits in a number are in ascending order:

```python
def is_tidy(number: int) -> bool:
    while number:
        last_digit = number % 10  # take last digit
        number = number // 10
        prelast_digit = number % 10  # take prelast digit
        if last_digit < prelast_digit:
            return False
    return True
```

```python
print(is_tidy(12))  # True
print(is_tidy(32))  # False
print(is_tidy(1024))  # False
print(is_tidy(3445))  # True
print(is_tidy(13579))  # True
```

```python
def is_tidy(number: int) -> bool:
    while number > 0:
        curr_number = number % 10  # current (the last) number
        number = number // 10  # the rest of a number
        if curr_number < number % 10:  # compare the current number and pre last number(the last in updated number)
            return False
    return True
```

```python
print(is_tidy(12))  # True
print(is_tidy(32))  # False
print(is_tidy(1024))  # False
print(is_tidy(3445))  # True
print(is_tidy(13579))  # True
```
