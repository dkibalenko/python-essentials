```python
# check if input string letters are in the same order with alphabet, and no letters missed in there
```

```python
def is_alphabet(letters: str) -> bool:
    letters = letters.lower()
    for i in range(len(letters) - 1):
        if ord(letters[i + 1]) - ord(letters[i]) != 1:
            return False
    return True
```

```python
print(is_alphabet("abc"))  # 'True'
print(is_alphabet("aBc"))  # 'True'
print(is_alphabet("abd"))  # 'False'
print(is_alphabet("a"))  # 'True'
print(is_alphabet("abcdefghjiklmnopqrstuvwxyz"))  # 'False'
print(is_alphabet("tuvwxyz"))  # 'True'
print(is_alphabet("XYZ"))  # 'True'
print(is_alphabet("mnoprqst"))  # 'False'
```
