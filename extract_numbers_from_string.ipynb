```python
def calculate_guests(guests_input: str):
    # write your code here
    new_data = ''.join((ch if ch in "0123456789." else " ") for ch in guests_input)
    if "." in new_data:
        return int(new_data.strip().split(".")[0])
    result = new_data.strip(' ')
    if not result or guests_input == "0":
        return "not a number"
    
    return int(result.split(" ")[0])
```

```python
print(calculate_guests("I think 5 guests"))
print(calculate_guests("Big company of 15 dudes"))
print(calculate_guests("5"))
print(calculate_guests("alone"))
print(calculate_guests("0"))
print(calculate_guests("Hello, 9.75 people"))
print(calculate_guests("There will be 7 or 9 guys"))
print(calculate_guests("hello cat walks on my keyboard ksadjfhskaj12.34kasdfhsjk"))
```

```python
def calculate_guests(guests_input: str):
    number_str = ""
    for char in guests_input:
        if char.isnumeric():  # test char if it's a number (positive 1-9)
            number_str += char  # add numbers as far as they are in a row
        elif len(number_str) != 0:  # once we get not a number char we break loop
            break
    return (
        int(number_str)
        if len(number_str) > 0 and int(number_str) != 0
        else "not a number"
    )
```

```python
print(calculate_guests("I think 5 guests"))
print(calculate_guests("Big company of 15 dudes"))
print(calculate_guests("5"))
print(calculate_guests("alone"))
print(calculate_guests("0"))
print(calculate_guests("Hello, 9.75 people"))
print(calculate_guests("There will be 7 or 9 guys"))
print(calculate_guests("hello cat walks on my keyboard ksadjfhskaj12.34kasdfhsjk"))
```
