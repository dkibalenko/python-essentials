```python
def get_years(amount: int, percent: int, limit: int) -> int:
    total_pay = amount
    count_pay_years = 0
    while True:
        total_pay += total_pay * (percent / 100)
        if total_pay > limit:
            break
        count_pay_years += 1
    return count_pay_years

print(get_years(1600, 10, 2000))
print(get_years(500, 3, 550))
print(get_years(5000, 5, 6500))
```

```python
def get_years(amount: int, percent: int, limit: int) -> int:
    years = 0
    coefficient = 1 + percent / 100  # annual pay, based on current total pay amount
    while amount * coefficient <= limit:
        amount *= coefficient
        years += 1
    return years
```

```python
print(get_years(1600, 10, 2000))
print(get_years(500, 3, 550))
print(get_years(5000, 5, 6500))
```
