### try...except...else...finally
Order of execution:
- `try`
- `except` (if error)
- `else` (if no error)
- `finally` (always)

---

### 1. finally always runs — even if you `return` inside try/except

```python
def f():
    try:
        return 1
    finally:
        return 2
```

```python
f()
```

Result: 2, because finally overrides everything.

---

### 2. finally runs on exception — even if exception is not caught

```python
try:
    1 / 0
finally:
    print("cleanup")
```

Then the exception propagates.

---

### 3. finally runs on `break`/`continue`
Even if you break or continue inside a loop:

```python
for i in range(3):
    try:
        break
    finally:
        print("runs anyway")
```

### 4. finally runs on sys exit
Even `sys.exit()` triggers finally.

```python
import sys
try:
    sys.exit()
finally:
    print("still runs")
```

### 5. finally can suppress exceptions
If `finally` raises a new exception, it **overwrites the original one**.

```python
try:
    1 / 0
finally:
    raise ValueError("new error")
```

The `ZeroDivisionError` is lost.

---

### 6. except cannot catch finally errors
If an exception happens in `finally`, it bypasses the `except` block.

```python
try:
    1 / 0
except ZeroDivisionError:
    print("caught")
finally:
    raise RuntimeError("boom")
```

### 7. return in finally is dangerous
Returning inside `finally` suppresses exceptions:

```python
def f():
    try:
        1 / 0
    finally:
        return "safe"
```

```python
f()
```

No exception is raised — the function returns `"safe"`.

This is considered bad practice and interviewers expect you to know why.
