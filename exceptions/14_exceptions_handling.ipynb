### Examples

```python
def give_cookies(count_of_cookies, count_of_children):
    return count_of_cookies // count_of_children
```

```python
give_cookies(10, 5)
```

```python
def give_cookies(count_of_cookies, children, favorite_number):
    count_of_cookies -= 10  # save for favorite child
    count_cookies_for_each_child = count_of_cookies // len(children)
    
    for i in range(len(children)):
        children[i] += count_cookies_for_each_child
        
    children[favorite_number] += 10
    
    print("Cookies for children", children)
```

```python
give_cookies(count_of_cookies=15, 
             children=[1, 0, 0, 1, 2], 
             favorite_number=3)
```

```python
give_cookies(count_of_cookies=15, 
             children=[], 
             favorite_number=3)
```

```python
try:
    give_cookies(count_of_cookies=15, 
                 children=[], 
                 favorite_number=3)
except Exception:
    print("So sad, no one wants cookies")
```

```python
try:
    give_cookies(count_of_cookies=15, 
                 children=[0, 0, 0, 0, 0], 
                 favorite_number=7)
except Exception:
    print("So sad, no one wants cookies")
```

### `try... except` usage

```python
def divide(a, b):
    return a // b

def get_element_by_index(arr, index):
    return arr[index]

def perform_calculation():
    a = 18
    b = 6
    
    try:
        index = divide(a, b)
    except ZeroDivisionError:
        print("There was a ZerodivisionError. Returning -1.")
        return -1
        
    arr = [1, 2, 3, 4, 5]
    
    try:
        return get_element_by_index(arr, index)
    except IndexError:
        print("There was an IndexError. Returning -1.")
        return -1
```

```python
print("Result:", perform_calculation())
```

### Raise exceptions, `assert`

```python
def get_driver_license(person):
    if person["age"] < 16:
        raise Exception("People under 16 years old are not allowed.")
        
    print("Successfully got driver license.")
```

```python
print(get_driver_license({
    "name": "John",
    "age": 15
}))
```

```python
try:
    print(get_driver_license({
    "name": "John",
    "age": 15
    }))
except Exception as e:
    print(e)
```

```python
def get_driver_license(person):
    assert person["age"] >= 16
        
    print("Successfully got driver license.")
```

```python
get_driver_license({
    "name": "John",
    "age": 15
    })
```

if it had been 25 years old it would have been accepted:

```python
get_driver_license({
    "name": "John",
    "age": 25
    })
```

but we have to supply our ***assert*** statement with some message

```python
def get_driver_license(person):
    assert person["age"] >= 16, "People under 16 years old are not allowed."
        
    print("Successfully got driver license.")
```

```python
get_driver_license({
    "name": "John",
    "age": 15
    })
```

keep in mind that `assert` statement always gives `AssertionError`

### Custom Exceptions

```python
class CustomError(Exception):
    """Our CustomError Exception"""
    pass
```

```python
raise CustomError("Hello from CustomError")
```

```python
event = "customerror"

try:
    if event == "custom error":
        raise CustomError("Raising & Handling CustomError.")
except CustomError as e:
    print(e)
```

### Examples

```python
class ValueTooBigError(CustomError):
    """This exception is raised when value is > 10."""
```

```python
class ValueTooSmallError(CustomError):
    """This exception is raised when value is < 5."""
```

```python
value = 12
```

```python
try:
    if value > 10:
        raise ValueTooBigError(f"{value} is too big")
    if value < 5:
        raise ValueTooSmallError(f"{value} is too small")
        
    print(f"value {value} is ok.")
except CustomError as e:
    print("Handled error:")
    print(e)
```

```python
class ValueTooBigErrorCustomizedError(Exception):
    """This customized exception will be raised when value is > 10."""
    def __init__(self, message="Value should be <= 10"):
        super().__init__(message)
        
        
value = 12

if value > 10:
    raise ValueTooBigErrorCustomizedError
```
