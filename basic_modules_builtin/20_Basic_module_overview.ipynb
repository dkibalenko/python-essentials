### `pickle`

```python
import pickle
```

```python
class User:
    def __init__(self, name, surname, age):
        self.name = name
        self.surname = surname
        self.age = age
        
users = [
    User(name="John", surname="Smith", age=30),
    User(name="Ana", surname="White", age=33)
]
```

```python
users
```

```python
with open("data.pickle", "wb") as f:
    pickle.dump(users, f)  # serialize(reformat) User objects and write the object to f variable
```

```python
del users
```

```python
users
```

```python
with open("data.pickle", "rb") as f:
    users = pickle.load(f)
```

```python
users
```

first we created Users objects and wrote them in binary mode to a `data.pickle` file, then we deleted User object from the memory, and deserialized `data.pickle` file data into Python User objects, with new memory addresss

```python
class User:
    def __init__(self, name, surname, age, best_friend):
        self.name = name
        self.surname = surname
        self.age = age
        self.best_friend = best_friend
        
user = User(
    name="John",
    surname="Smith",
    age=25,
    best_friend=User(
            name="Kate",
            surname="Black",
            age=22,
            best_friend=User(
                name="Bob",
                surname="White",
                age=30,
                best_friend=None)))

print(vars(user))
print(vars(user.best_friend))
print(vars(user.best_friend.best_friend))
```

```python
with open("data.pickle", "wb") as f:
    pickle.dump(user, f)
```

```python
del user
```

```python
with open("data.pickle", "rb") as f:
    user = pickle.load(f)
```

```python
print(vars(user))
print(vars(user.best_friend))
print(vars(user.best_friend.best_friend))
```

### `dataclasses`

```python
class User:
    def __init__(self, name, surname, age, best_friend):
        self.name = name
        self.surname = surname
        self.age = age
```

another way of creating classes without repetition code

```python
@dataclasses.dataclass
class User:
    name: str
    surname: str
    age: int
```

decorator `dataclass` will create `__init__` constructor with all data

```python
@dataclass
class Book:
    name: str
    author: str
        
@dataclass
class BookShelf:
    books: list[Book]
book_shelf = BookShelf([Book(name="Harry Potter", author="J. Rowlig")])
```

### practice

```python
import dataclasses
from datetime import datetime
import pickle


@dataclasses.dataclass
class Specialty:
    name: str
    number: int | float


@dataclasses.dataclass
class Student:
    first_name: str
    last_name: str
    birth_date: datetime
    average_mark: float
    has_scholarship: bool
    phone_number: str
    address: str


@dataclasses.dataclass
class Group:
    specialty: Specialty
    course: str
    students: list[Student]


def write_groups_information(groups: list[Group]):
    with open("groups.pickle", "wb") as pickle_groups:
        pickle.dump(groups, pickle_groups, protocol=pickle.HIGHEST_PROTOCOL)

    # return the maximum number of students from all the groups
    groups = sum(len(group.students) for group in groups)
    return groups


def write_students_information(students: list[Student]):
    with open("students.pickle", "wb") as pickle_students:
        pickle.dump(students, pickle_students, protocol=pickle.HIGHEST_PROTOCOL)

    # return the number of students
    return len(students)


def read_groups_information():
    with open("groups.pickle", "rb") as pickle_groups:
        groups = pickle.load(pickle_groups)

    return [group.specialty.name for group in groups]


def read_students_information() -> list:
    with open("students.pickle", "rb") as pickle_students:
        students = pickle.load(pickle_students)

    return students
```

```python
student1 = Student(
    first_name="John",
    last_name="Smith",
    birth_date=datetime(1987, 7, 3),
    average_mark=90.5,
    has_scholarship=True,
    phone_number="+380660011302",
    address="Kherson")

student2 = Student(
    first_name="Alise",
    last_name="White",
    birth_date=datetime(1945, 6, 23),
    average_mark=44.9,
    has_scholarship=True,
    phone_number="+12575757575",
    address="Baltimore")

student3 = Student(
    first_name="Bobic",
    last_name="Fucker",
    birth_date=datetime(1999, 3, 12),
    average_mark=56.4,
    has_scholarship=True,
    phone_number="+494848488442",
    address="Greenland")

student4 = Student(
    first_name="Kolya",
    last_name="Lukashenko",
    birth_date=datetime(1996, 3, 23),
    average_mark=1.2,
    has_scholarship=False,
    phone_number="+848484848444",
    address="Minsk")
```

```python
specialty1 = Specialty(name="Full Stack Developer", number=1010)
```

```python
specialty2 = Specialty(name="Q&A", number=94949)
```

```python
group1 = Group(
    specialty=specialty1,
    course="49/2024",
    students=[student1, student2])
```

```python
group2 = Group(
    specialty=specialty2,
    course="949/2023",
    students=[student3, student4])
```

```python
write_groups_information([group1, group2])
```
