**Demo packages** (`demo_packages/`):

|Package|What it demonstrates|
|---|---|
|`lifecycle_demo/once.py`|Print-traces that appear _exactly once_ regardless of how many importers. `_call_count` is global shared state.|
|`lifecycle_demo/importer1.py` + `importer2.py`|Two modules importing the same thing — shows single-init in practice|
|`pkg_a/` (with `sub/`)|Nested package, `__init__.py` exposure, `__all__`, intra-package imports|
|`pkg_b/consumer.py`|Cross-package import — `pkg_a.__init__` does NOT re-fire|
|`circular_demo/broken_a+b.py`|The exact ImportError, left intentionally broken for study|
|`circular_demo/fixed_a+b.py`|Fix 1: deferred import inside function body|
|`circular_demo/module_ref_a+b.py`|Fix 2: import the module object, not the symbol|
