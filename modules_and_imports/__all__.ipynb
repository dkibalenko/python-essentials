```py
from movies.routers.crud import router as crud_router
from movies.routers.genres import router as genres_router
from movies.routers.interactions import router as interactions_router
from movies.routers.video import router as video_router

__all__ = [
    "crud_router",
    "genres_router",
    "interactions_router",
    "video_router",
]
```

`__init__.py` runs when the **package itself** is first imported — `import movies.routers` or any submodule import like `from movies.routers import crud_router`. Both trigger `__init__.py`.

1. **Re-exports with aliases** — `from movies.routers.crud import router as crud_router` pulls the `router` object from the submodule and binds it to the name `crud_router` in the package namespace. Every submodule has its own `router`, so the aliases disambiguate them.
    
2. **Flattens the public API** — consumers write `from movies.routers import crud_router` instead of `from movies.routers.crud import router`. The internal file structure becomes an implementation detail.
    
3. **`__all__`** — defines what gets exported when someone does `from movies.routers import *`. It also serves as explicit documentation of the intended public interface, and tools like mypy and IDEs use it to know what the package intentionally exposes.
    

The one nuance: `__all__` does **not** prevent direct imports of things not listed in it. `from movies.routers.crud import router` still works fine — `__all__` only gates the `import *` behavior, not access in general.
