**Adding condition**

Get all the even numbers from 0 - 30

```python
evens = []

for num in range(30):
    is_even = num % 2 == 0
    if is_even:
        evens.append(num)

evens
```

```python
evens = [num for num in range(30) if num % 2 == 0]

evens
```

**Comprehension with Multiple conditions**

Strings that start with 'a' and end in 'y'

```python
options = ["any", "albany", "apple", "world", "hello", ""]
valid_strings = []

for s in options:
    if len(s) <= 1:
        continue

    if s[0] != "a":
        continue

    if s[-1] != "y":
        continue

    valid_strings.append(s)

valid_strings
```

```python
valid_strings = [
    s
    for s in options
    if len(s) >= 2    # add s if it's more than 2 chars long
    if s[0] == "a"    # AND if s starts with 'a'
    if s[-1] == "y"   # AND if s ends with 'y'
]

valid_strings
```

**Multiple list comprehensions**

Flattening a matrix (list of lists)

```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flattened = []

for row in matrix:             # outer loop
    for num in row:            # inner loop
        flattened.append(num)  # value to add

flattened
```

```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

flattened = [
    num                 # value to add
    for row in matrix   # outer most loop
    for num in row      # inner loop
]

flattened
```

**`if/else` in a comprehension**

Categorize numbers as even or odd

```python
categories = []

for num in range(10):
    if num % 2 == 0:
        categories.append("Even")
    else:
        categories.append("Odd")

categories
```

```python
categories = [
    "Even"                     # 1. if branch value
    if num % 2 == 0
    else "Odd"                 # 2. else branch value
    for num in range(10)       # 3. for loop
]

categories
```

**Nested List comprehension**

Build a 3D list

```python
import pprint

printer = pprint.PrettyPrinter()

lst = []

for a in range(5):
    l1 = []
    for b in range(5):
        l2 = []
        for num in range(5):
            l2.append(num)
        l1.append(l2)
    lst.append(l1)

printer.pprint(lst)
```

```python
_3d_list = [
    [
        [
            num for num in range(5)     # inner most for loop
        ]
        for b in range(5)               # inner list for loop
    ]
    for a in range(5)                   # main list for loop
]

_3d_list
```

**Set Comprehension**

Removing duplicates from a list while applying a function

```python
nums = [1, 2, 3, 4, 5, 6, 7, 8, 9, 2, 4, 3, 5, 9, 9, 8]

unique_squares = {x**2 for x in nums}

unique_squares
```

**Generator Comprehension - Generator Expression**

`genexpr = (expression for item in collection)`

Generate the sum of all the squares from the numbers 0 to 1_000_000

```python
sum_of_squares = sum(x**2 for x in range(1_000_000))
```
