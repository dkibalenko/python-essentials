```python
import random
```

### random.random() returns random number from 0-1(1 is excl)

```python
random.random()
```

### generate random numbers from 0 through 5 (5 is excl)

```python
for i in range(10):
    x = random.random() * 5
    print(x)
```

### random numbers from 2 through 7:

```python
for i in range(10):
    x = random.random() * 5 + 2
    print(x)
```

### by adding 2, we added 2 to lower bound, and 2 to upper bound, since we got 2 - 7

```python
for i in range(10):
    x = int(random.random() * 5 + 2)  # only integer
    print(x)
```

```python
for i in range(10):
    x = random.randint(2, 6)  # integer form 2 to incl 6
    print(x)
```
