### Iterators Example

```python
class EachSecondElement:
    def __init__(self, elements):
        self.elements = elements
        
    def __iter__(self):
        self.current_element = 1
        return self
    
    def __next__(self):
        if self.current_element >= len(self.elements):
            raise StopIteration
            
        result = self.elements[self.current_element]
        self.current_element += 2
        return result
```

```python
students = ["Alice", "Bob", "Carl", "John"]
```

```python
each_second_person = EachSecondElement(students)
```

```python
iterator = iter(each_second_person)
```

```python
next(iterator)
```

```python
next(iterator)
```

```python
next(iterator)
```

```python
for student in EachSecondElement(students):
    print(student)
```

```python
for student in EachSecondElement("hellooooo"):
    print(student)
```

***Cyclic iterator***

```python
class EachSecondElementCyclic:
    def __init__(self, elements):
        self.elements = elements
        
    def __iter__(self):
        self.current_element = 1
        return self
    
    def __next__(self):
        result = self.elements[self.current_element]
        self.current_element += 2
        self.current_element %= len(self.elements)
        return result
```

```python
students = ["Alice", "Bob", "Carl", "John", "Kate"]
```

```python
each_second_person_cyclic = EachSecondElementCyclic(students)
```

```python
iterator_cyclic = iter(each_second_person_cyclic)
```

```python
for _ in range(5):
    print(next(iterator_cyclic))
```

```python
def add_subtract_multiply_divide(num: int):
    start = 10
    
    after_addition = start + num
    print("after_addition")
    yield after_addition
    
    after_subtraction = after_addition - num
    print("after_subtraction")
    yield after_subtraction
    
    after_multiplication = after_subtraction * num
    print("after_multiplication")
    yield after_multiplication
    
    after_division = after_multiplication / num
    print("after_division")
    yield after_division
```

```python
add_subtr_mul_div_with_5 = add_subtract_multiply_divide(5)
```

```python
next(add_subtr_mul_div_with_5)
```

### Iterators vs Generators

```python
class EachSecondElement:
    def __init__(self, elements):
        self.elements = elements
        
    def __iter__(self):
        self.current_element = 1
        return self
    
    def __next__(self):
        if self.current_element >= len(self.elements):
            raise StopIteration
            
        result = self.elements[self.current_element]
        self.current_element += 2
        return result
```

```python
def each_second_element(elements):
    current_element = 1
    
    while current_element < len(elements):
        yield elements[current_element]
        current_element += 2
```

```python
students = ["Alice", "Bob", "Carl", "John", "Kate"]
```

```python
for element in EachSecondElement(students):
    print(element)
```

```python
for element in each_second_element(students):
    print(element)
```
