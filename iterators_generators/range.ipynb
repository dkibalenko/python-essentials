### `range()` vs `enumerate()`

`range()` is a built-in function used to **iterate through a sequence of numbers**. 

`enumerate()` is a built-in function to iterate through a sequence and keep track of `both` the `index` and the `number`.

```python
type(range)
```

```python
type(range(-10, 0))
```

```python
list(range(-10, 0))
```

```python
list(range(4, 2))
```

```python
list(range(20, 0, -2))
```

```python
list(range(5, -1, -1))
```

### `range` Real-World Example
Say that you want to generate a sequence of numbers representing the leap years between 2000 and 2100:

```python
leap_years = range(2000, 2100, 4)
```

```python
list(leap_years)
```
