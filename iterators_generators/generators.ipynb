**Generator Function**

```python
def fib(num: int) -> int:
    a, b = 1, 1
    for _ in range(num):
        yield a  # generate the whole fib sequence
        a, b = b, a + b
```

```python
fib_gen = fib(5)
```

```python
for item in fib_gen:
    print(item)
```

**Generator Expression**

```python
odd_squares = (x**2 for x in range(1, 10) if x % 2 != 0)
```

```python
for square in odd_squares:
	print(square)
```
