**1. The simplest possible version (`range(stop)`)**

```python
# Python's range() is an ITERABLE
r = range(4)
print(hasattr(r, "__iter__"))  # only returns interator
print(hasattr(r, "__next__"))  # no iterator protocol
```

```python
def my_range(stop: int) -> int:
    i = 0
    while i < stop:
        yield i
        i += 1
```

- Start at 0
- Yield numbers one by one
- Stop when reaching stop

This is the **core idea** behind Python’s real range.

```python
obj = my_range(5)  # returns a generator object (which is iterator)
type(obj)
```

```python
hasattr(obj, "__iter__")
```

```python
hasattr(obj, "__next__")
```

**2. Add start and stop (`range(start, stop)`)**

```python
def my_range(start, stop):
    i = start
    while i < stop:
        yield i
        i += 1
```

```python
gen = my_range(1, 6)
```

```python
for item in gen:
    print(item)
```

```python
for item in gen:  # this time it's exhausted
    print(item)
```

```python
next(gen)
```

**3. Full version with step (`range(start, stop, step)`)**

```python
def my_range(start, stop=None, step=1):
    # handle one param: range(5) -> 0 - 4
    # handle two params: range(1, 5) -> 1 - 4
    # if step is given: postitive or negative or zero(raise)?
    # handle three params: range(10, 0, -2) -> 10 - 1 over 2 backwards move

    # Handle range(stop)
    if stop is None:
        stop = start
        start = 0

    i = start

    if step > 0:               # Positive step
        while i < stop:
            yield i
            i += step
    elif step < 0:             # Negative step
        while i > stop:
            yield i
            i += step
    else:
        raise ValueError("step cannot be zero")
```

```python
for x in my_range(5):
    print(x)
```

```python
for x in my_range(2, 7):
    print(x)
```

```python
for x in my_range(10, 0, -2):
    print(x)
```
